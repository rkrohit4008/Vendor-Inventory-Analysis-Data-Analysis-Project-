```python
import pandas as pd
from sqlalchemy import create_engine
import os
```

```python
engine = create_engine('sqlite:///inventory.db')
```

```python
def ingest_db(df,table_name,engine):
    # ''' this function will ingest the dataframe into database table '''
    df.to_sql(table_name,con = engine,if_exists = 'replace', index = False)
```

```python
def load_raw_data():
    # ''' this function will load the CSV as dataframe and ingest into db'''
    data_path = r'D:\Project_1'
    for file in os.listdir(data_path):
        if '.csv' in file:       # To ignore".pynb" file
            print(file)
            file_path = os.path.join(data_path, file)   #To find the csv file with path we have to join the path with file name
            df = pd.read_csv(file_path)
            print(df.shape)
            ingest_db(df,file[:-4],engine)   #to insert the updated data in db apply function which is created

load_raw_data()
```
